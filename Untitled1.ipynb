```python
# Import required libraries
import pandas as pd

# 1. Load the 2019 dataset
# Ensure 'superstore_2019.csv' is in the same folder as this new notebook
df_2019 = pd.read_csv('superstore_2019.csv')

# 2. Display column names, data types, and general info
print("--- 2019 Data: General Information & Data Types ---")
print(df_2019.info())

# 3. Display dataset shape (Rows, Columns)
print("\n--- 2019 Data: Shape of Dataset (Rows, Columns) ---")
print(df_2019.shape)

# 4. Count missing values per column
print("\n--- 2019 Data: Missing Values Count Per Column ---")
print(df_2019.isnull().sum())

# 5. Count total duplicate rows
print("\n--- 2019 Data: Total Duplicate Rows ---")
print(df_2019.duplicated().sum())

# 6. Preview the first 5 rows
print("\n--- 2019 Data: Preview First 5 Rows ---")
df_2019.head()
```

```python
# Import required libraries
import pandas as pd

# 1. Drop duplicate rows
df_2019 = df_2019.drop_duplicates()

# 2. Fix Data Formats for all 19 columns
# Convert Date columns to datetime format
df_2019['order_date'] = pd.to_datetime(df_2019['order_date'])
df_2019['ship_date'] = pd.to_datetime(df_2019['ship_date'])

# Convert Text/Categorical columns to string type
text_columns = [
    'order_id', 'customer', 'manufactory', 'product_name', 
    'segment', 'category', 'subcategory', 'region', 
    'city', 'state', 'country'
]
for col in text_columns:
    df_2019[col] = df_2019[col].astype(str)

# Convert Numeric columns to correct numeric types (Integer/Float)
df_2019['zip'] = df_2019['zip'].astype(int)
df_2019['quantity'] = df_2019['quantity'].astype(int)

df_2019['discount'] = df_2019['discount'].astype(float)
df_2019['profit'] = df_2019['profit'].astype(float)
df_2019['sales'] = df_2019['sales'].astype(float)
df_2019['profit_margin'] = df_2019['profit_margin'].astype(float)

# 3. Final Check on the Cleaned Data
print("--- Final Shape of Cleaned Dataset ---")
print(df_2019.shape)

print("\n--- Final Data Types Check ---")
print(df_2019.dtypes)

print("\n--- Final Missing Values Check ---")
print(df_2019.isnull().sum().sum())

print("\n--- Final Duplicate Rows Check ---")
print(df_2019.duplicated().sum())
```

```python
# Save the cleaned 2019 dataset to a new CSV file
df_2019.to_csv('superstore_2019_cleaned.csv', index=False)
print("File 'superstore_2019_cleaned.csv' has been saved successfully!")
```

```python
# --- Page 1: Overview Analysis ---

# 1. Calculate Main KPIs (Business Logic)
total_sales = df_2019['sales'].sum()
total_profit = df_2019['profit'].sum()
total_quantity = df_2019['quantity'].sum()

# Correct Concept for Overall Profit Margin: (Total Profit / Total Sales)
overall_profit_margin = (total_profit / total_sales) * 100

print("--- Main KPIs for 2019 ---")
print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Total Quantity Sold: {total_quantity:,} units")
print(f"Overall Profit Margin: {overall_profit_margin:.2f}%")

# 2. Monthly Trend Analysis (Sales & Profit by Month)
# Extract Month number and Month name for sorting and display
df_2019['month_num'] = df_2019['order_date'].dt.month
df_2019['month_name'] = df_2019['order_date'].dt.strftime('%B')

# Group by month and calculate sum of sales and profit
monthly_trend = df_2019.groupby(['month_num', 'month_name'])[['sales', 'profit']].sum().reset_index()

print("\n--- Monthly Sales & Profit Trend ---")
print(monthly_trend)
```

```python
# 3. Analyze Sales and Profit by Category
category_analysis = df_2019.groupby('category')[['sales', 'profit']].sum().reset_index()

# 4. Analyze Sales and Profit by Segment (Customer Type)
segment_analysis = df_2019.groupby('segment')[['sales', 'profit']].sum().reset_index()

print("--- Sales & Profit by Product Category ---")
print(category_analysis)

print("\n--- Sales & Profit by Customer Segment ---")
print(segment_analysis)
```

# 📊 Executive Summary & Strategic Data Story: The 2019 Discount Trap

### 📖 The Story Behind the Numbers: The Danger of Aggressive Discounting
In 2019, our Superstore achieved **$484,216.03 in Total Sales** and **$49,543.97 in Net Profit** with an overall **10.23% Profit Margin**. Operatively, our logistics network maintained a consistent **4.00-day average shipping duration**. 

However, our data reveals a dangerous reality: **Our unmonitored Discount policy (`discount`) is the single core catalyst behind every financial leak in the company.**

---

### 🚨 Critical Performance Bottlenecks (Proven by Data)

#### 1. The Product Mirage: Tables & Bookcases Discount Crisis
* **The Data:** **Tables** are our 2nd highest revenue sub-category ($46,888.36 sales) but generated a catastrophic net loss of **-$3,124.04**. 
* **The Discount Factor:** The data proves we are giving an average **26.6% discount rate on Tables** and **20.5% on Bookcases**. This aggressive pricing, combined with the heavy weight and shipping costs of furniture, completely killed our profit margins.

#### 2. The Geographical Trap: The Texas Discount Disaster
* **The Data:** **Texas** is our 3rd largest state by revenue ($50,625.17 sales) but generated a devastating net loss of **-$9,135.01**.
* **The Discount Factor:** Texas has a shocking average discount rate of **36.9%** across all orders! In comparison, **California** maintains a low average discount of **7.1%**, which allowed it to generate our highest net profit of **$12,637.94**.

#### 3. The Customer Illusion: Sean Miller's Account Exploitation
* **The Data:** **Sean Miller** is our #1 buyer by volume ($23,661.22 sales) but cost the company a net loss of **-$1,789.73**.
* **The Discount Factor:** Sean Miller maintained a high average discount profile of **28.5%** on his purchases. He strategically exploited our promo system by targeting highly discounted, low-margin items.

---

### 🎯 Strategic Recommendations (Data-Driven Action Plan)

* **🛑 Enforce a Strict 20% Discount Cap:** Immediately restrict the checkout system to a **20% maximum discount limit per order** to stop margin manipulation by high-volume buyers like Sean Miller.
* **📉 Eliminate Discounts on Heavy Furniture:** Completely remove promotional discounts on **Tables and Bookcases**, especially in long-distance states where shipping costs are high.
* **📍 Shift Marketing Budget Away from Texas:** Cease all discount-heavy promotional campaigns in Texas. Reallocate marketing budget to high-margin, low-discount states like **California** (7.1% avg discount) and **New York** (5.3% avg discount) to maximize corporate ROI.
